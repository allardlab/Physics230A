# Problem Set 0: Scavenger hunt

Physics 230A/146A — Biological Physics

Jun Allard

Find values for the following biological quantities. You may use Google,
Wikipedia, textbooks, peer-reviewed papers, large language models, or
anything else, but you must cite your sources[1]. One valuable source
is: `bionumbers.org`. In cases where answers need to be qualified, do
so. Are there ranges? Are there a few exceptions or many exceptions? Are
these order-of-magnitude estimates?

1.  The number of amino acids in the average human protein

The size of (in meters or other SI unit):

1.  Longest nerves in humans
2.  A eukaryotic cell
3.  A bacterial cell
4.  An organelle, like the eukaryotic nucleus
5.  The width of the plasma membrane
6.  Your favorite protein
7.  The resolution of a traditional light microscope
8.  The resolution of a traditional electron microscope
9.  A water molecule
10. The width of DNA

The time for (in seconds or other SI unit):

1.  Turnover of cells in the human body
2.  The generation time (time between divisions) of a bacterium in lab
    conditions
3.  Transcription of one DNA basepair to RNA
4.  Translation of one amino acid in an RNA into protein
5.  Human reflex
6.  Protein folding

The energy of:

1.  The reaction ATP $\rightarrow$ ADP-Pi
2.  The thermal energy unit $k_B T$ (where $k_B$ is Boltzmann’s constant
    and $T$ is the temperature from absolute zero)
3.  Moving one electron against the electric field across the membrane
    of a cell

[1] and those sources must have provenance (i.e., they can cite their
sources)
